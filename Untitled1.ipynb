# 1.Problem Statement Analysis
Maintaining academic details manually can become difficult as the number of records increases. I am creating a menu-driven Python program to manage the details in one place.

INPUTS: roll number, registration number, name, department, date of birth, subjects, marks, attendance, email, and address.

OUTPUTS: complete record details, search/update/delete results, average marks, highest scorer, records by department, and total record count.

ASSUMPTIONS: roll numbers are unique; marks and attendance are percentages from 0 to 100; information remains in memory while the program runs and is not saved permanently.

# 2. Data Structure Selection

i] String     -               	Name, department, email, address, and date of birth

ii] List    -                	Subjects, marks, and attendance

iii] Tuple        -            	Fixed identity details such as roll number, registration number, and date of birth

iv] Set     -                   	Unique department and subject names

v] Dictionary         -      	Complete details for each record

```python
# Student records will be stored in this list
students = []

print("Student Record Management System Started")
```

```python
def add_student():
    print("\n--- Add Student ---")

    roll_number = int(input("Enter Roll Number: "))

    # Check for duplicate roll number
    for student in students:
        if student["Roll Number"] == roll_number:
            print("Student already exists!")
            return

    name = input("Enter Name: ")
    department = input("Enter Department: ")
    registration_number = input("Enter Registration Number: ")
    date_of_birth = input("Enter Date of Birth: ")
    email = input("Enter Email: ")
    phone = input("Enter Phone Number: ")

    subjects = input("Enter subjects separated by comma: ").split(",")
    subjects = [subject.strip() for subject in subjects]

    marks_input = input("Enter marks separated by comma: ").split(",")
    marks = [float(mark.strip()) for mark in marks_input]

    attendance_input = input("Enter attendance separated by comma: ").split(",")
    attendance = [float(att.strip()) for att in attendance_input]

    clubs = input("Enter student clubs separated by comma: ").split(",")
    clubs = {club.strip() for club in clubs}

    student = {
        "Roll Number": roll_number,
        "Registration Number": registration_number,
        "Date of Birth": date_of_birth,
        "Name": name,
        "Department": department,
        "Subjects": subjects,
        "Marks": marks,
        "Attendance": attendance,
        "Contact Information": {
            "Email": email,
            "Phone": phone
        },
        "Clubs": clubs
    }

    students.append(student)

    print("Student added successfully!")
```

```python
def search_student():
    print("\n--- Search Student ---")
    
    roll_number = int(input("Enter Roll Number to search: "))
    
    for student in students:
        if student["Roll Number"] == roll_number:
            print("\nStudent Found!")
            print("Roll Number:", student["Roll Number"])
            print("Name:", student["Name"])
            print("Department:", student["Department"])
            print("Registration Number:", student["Registration Number"])
            print("Date of Birth:", student["Date of Birth"])
            print("Subjects:", student["Subjects"])
            print("Marks:", student["Marks"])
            print("Attendance:", student["Attendance"])
            print("Email:", student["Contact Information"]["Email"])
            print("Phone:", student["Contact Information"]["Phone"])
            print("Clubs:", student["Clubs"])
            return
    
    print("Student not found!")
```

```python
def update_student():
    print("\n--- Update Student ---")
    
    roll_number = int(input("Enter Roll Number to update: "))
    
    for student in students:
        if student["Roll Number"] == roll_number:
            
            print("Student found.")
            
            new_name = input("Enter new Name: ")
            new_department = input("Enter new Department: ")
            new_email = input("Enter new Email: ")
            new_phone = input("Enter new Phone Number: ")
            
            student["Name"] = new_name
            student["Department"] = new_department
            student["Contact Information"]["Email"] = new_email
            student["Contact Information"]["Phone"] = new_phone
            
            print("Record updated successfully!")
            return
    
    print("Student not found!")
```

```python
def delete_student():
    print("\n--- Delete Student ---")
    
    roll_number = int(input("Enter Roll Number to delete: "))
    
    for student in students:
        if student["Roll Number"] == roll_number:
            students.remove(student)
            print("Record deleted successfully!")
            return
    
    print("Student not found!")
```

```python
def display_students():
    print("\n--- All Student Records ---")
    
    if len(students) == 0:
        print("No student records available.")
        return
    
    for student in students:
        print("\n----------------------------")
        print("Roll Number:", student["Roll Number"])
        print("Name:", student["Name"])
        print("Department:", student["Department"])
        print("Registration Number:", student["Registration Number"])
        print("Date of Birth:", student["Date of Birth"])
        print("Subjects:", student["Subjects"])
        print("Marks:", student["Marks"])
        print("Attendance:", student["Attendance"])
        print("Email:", student["Contact Information"]["Email"])
        print("Phone:", student["Contact Information"]["Phone"])
        print("Clubs:", student["Clubs"])
        print("----------------------------")
```

```python
def calculate_average():
    print("\n--- Calculate Average Marks ---")
    
    roll_number = int(input("Enter Roll Number: "))
    
    for student in students:
        if student["Roll Number"] == roll_number:
            marks = student["Marks"]
            
            if len(marks) == 0:
                print("No marks available.")
                return
            
            total = sum(marks)
            average = total / len(marks)
            
            print("Student Name:", student["Name"])
            print("Total Marks:", total)
            print("Average Marks:", round(average, 2))
            return
    
    print("Student not found!")
```

```python
def highest_scorer():
    print("\n--- Highest Scorer ---")
    
    if len(students) == 0:
        print("No student records available.")
        return
    
    highest_student = students[0]
    highest_average = sum(highest_student["Marks"]) / len(highest_student["Marks"])
    
    for student in students[1:]:
        if len(student["Marks"]) > 0:
            average = sum(student["Marks"]) / len(student["Marks"])
            
            if average > highest_average:
                highest_average = average
                highest_student = student
    
    print("Highest Scorer:", highest_student["Name"])
    print("Roll Number:", highest_student["Roll Number"])
    print("Department:", highest_student["Department"])
    print("Average Marks:", round(highest_average, 2))
```

```python
def list_by_department():
    print("\n--- Students by Department ---")
    
    department = input("Enter Department: ")
    found = False
    
    for student in students:
        if student["Department"].lower() == department.lower():
            print("Roll Number:", student["Roll Number"])
            print("Name:", student["Name"])
            print("Department:", student["Department"])
            print("----------------------------")
            found = True
    
    if not found:
        print("No students found in this department.")
```

```python
def count_students():
    print("\n--- Student Count ---")
    print("Total Number of Students:", len(students))
```

```python
# Main Menu
while True:
    print("\n========== STUDENT RECORD MANAGEMENT SYSTEM ==========")
    print("1. Add Student")
    print("2. Search Student")
    print("3. Update Student")
    print("4. Delete Student")
    print("5. Display Records")
    print("6. Calculate Average")
    print("7. Find Highest Scorer")
    print("8. List Students by Department")
    print("9. Count Students")
    print("10. Exit")

    choice = input("Enter your choice (1-10): ")

    if choice == "1":
        add_student()

    elif choice == "2":
        search_student()

    elif choice == "3":
        update_student()

    elif choice == "4":
        delete_student()

    elif choice == "5":
        display_students()

    elif choice == "6":
        calculate_average()

    elif choice == "7":
        highest_scorer()

    elif choice == "8":
        list_by_department()

    elif choice == "9":
        count_students()

    elif choice == "10":
        print("Thank you for using Student Record Management System!")
        break

    else:
        print("Invalid choice! Please enter a number from 1 to 10.")
```
